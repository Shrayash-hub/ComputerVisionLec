```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread("/content/dataset-card.jpg") # used to read the images
# cv2.cvtColor() converts one color format to another.
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB) # converting the image from BGR fromat to RGB format as openCv loads image in BGR format and matplotlib needs
# it RGB format
plt.imshow(img)
plt.title("Step1: RGB image")
plt.axis("off")
plt.show()
```

```python
print(img.shape) # (height, width, number of color channels)  3 channels = [Red, Green, Blue]
```

```python
print(img[0:5,0:5]) # prints the pixel values of a small part of the image -> img[row_start:row_end , col_start:col_end]
# extracting a 5 × 5 pixel block from the top-left corner of the image.
```

```python
gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY) # converting the image from RGB to GRAY format
# instead of storing 3 numbers per pixel, we store 1 number per pixel.
plt.imshow(gray, cmap="gray")
plt.title("Step2: GRAYSCALE image")
plt.axis("off")
plt.show()
# cmap stands for color map.Since gray contains only intensity values (0–255), we tell matplotlib to display them using the grayscale color map.
```

```python
print(gray[0:10,0:10]) # prints the pixel intensity values of the top-left 10 × 10 region of the grayscale image.
```

```python
sampled = cv2.resize(gray,(50,50)) # the new image will be: 50 pixels wide 50 pixels tall
# This process is called image sampling / downsampling.
# Reducing image size helps in: 1 faster computation, 2 machine learning preprocessing, 3 reducing memory usage, 4 feature extraction
plt.imshow(sampled, cmap="gray")
plt.title("Step3: Sampled image(Low Resulution)")
plt.axis("off")
plt.show()
```

```python
plt.figure(figsize=(4,4)) # Using Matplotlib to create a plotting window. figsize=(10,10) means: width = 10 inches height = 10 inches
# This makes the image large on the screen, which helps us see the grid and pixels clearly.

plt.imshow(sampled, cmap="gray")
# This shows the resized grayscale image (sampled).cmap="gray" tells matplotlib: low values → blackhigh values → white So the pixel intensity values are displayed correctly.
plt.grid(True) # This draws grid lines on the image. Because the image is 50×50, the grid helps visualize that the image is actually a matrix of pixels.
plt.title("Pixel, Grid")
plt.show()
```
