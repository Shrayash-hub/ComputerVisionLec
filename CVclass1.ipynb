```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread("/content/dataset-card.jpg") # used to read the images
# cv2.cvtColor() converts one color format to another.
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB) # converting the image from BGR fromat to RGB format as openCv loads image in BGR format and matplotlib needs
# it RGB format
plt.imshow(img)
plt.title("Step1: RGB image")
plt.axis("off")
plt.show()
```

```python
print(img.shape) # (height, width, number of color channels)  3 channels = [Red, Green, Blue]
```

```python
print(img[0:5,0:5]) # prints the pixel values of a small part of the image -> img[row_start:row_end , col_start:col_end]
# extracting a 5 × 5 pixel block from the top-left corner of the image.
```
